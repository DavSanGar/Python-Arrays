10.Cambia la forma del array para que tenga una estructura de tipo (filas, columnas)

```python
import numpy as np

longitud = int(input("Escribe la longitud del array: "))

array1 = np.ones(longitud)

print("El array queda así: ", array1)
print("---")

filas = int(input("Indica cuantas filas tiene el array: "))
columnas = int(input("Indica cuantas columnas tiene el array: "))

if filas * columnas == longitud:
    nuevo_array = np.reshape(array1, (filas, columnas))
    print("Este es el array dado con filas y columnas:\n ", nuevo_array)
else:
    print("Error en el numero de filas y/o columnas. La logitud de alguna de las dos no es compatible ")
```

12.Concatena ambas estructuras horizontalmente y verticalmente
(Pista: Investiga el funcionamiento de concatenate() y de vstack() y hstack() de numpy)

```python
import numpy as np

valores = int(input("Indica cuantos valores tiene el array: "))

array1 = np.ones(valores)

print("El array queda así:\n ", array1)
print("---")

array_remodelado = array1.reshape(4,4)

print("Array con reshape:\n ", array_remodelado)
print("---")

matriz_id = np.eye(4,4)
print("Esta es la matriz identidad:\n ",matriz_id)
print("---")

array_conca = np.concatenate((array_remodelado, matriz_id), axis=None)
print("Array con concatenate:\n ",array_conca)
print("---")

array_vstack = np.vstack((array_remodelado, matriz_id))
print("Array con vstack:\n ",array_vstack)
print("---")

array_hstack = np.hstack((array_remodelado, matriz_id))
print("Array con hstack:\n ",array_hstack)
print("---")
```
