4. Suma todos los elementos del array_2 usando un bucle y después usando un método
de numpy. Compara los resultados

```python
import numpy as np
array_2 = np.arange(2,11,2)
print(array_2)
print("---")

#---Suma de array con bucle
suma_total = 0
for i in array_2:
    suma_total = suma_total + i
print("Suma de los números con bucle",suma_total)

#---Suma en numpy con .sum
suma_array = array_2.sum()
print("Suma de los números con opción numpy",suma_array)
```

6.Encuentra los elementos comunes entre array_1 y array_2 y entre array_2 y array_2_revertido
(Pista: Investiga el uso de intersect1d() de numpy)

```python
import numpy as np
#array 1 con función 0 y cambiado a valores 2
array_1 = np.zeros(8)
array_1[:] = 2
print("Este es el array 1: ",array_1)

#array 2 con funcion de rango, del 2 al 10 con paso 2
array_2 = np.arange(2,11,2)
print("Este es el array 2: ",array_2)

#Invertir array 2
array_2_invertido = array_2[::-1]
print("Este es el array 2 invertido: ",array_2_invertido)

#Elementos comunes entre array 1 y 2
array1VSarray2 = np.intersect1d(array_1, array_2)
print("Estos son los valores comunes ",array1VSarray2)

#Elementos comunes entre array 2 y 2 invertido
array2VSarray2_inv = np.intersect1d(array_2, array_2_invertido)
print("Estos son los valores comunes ",array2VSarray2_inv)
```
