2. Multiplica todos los elementos del array usando un bucle y después usando un
método de numpy. Compara los resultados

```python
import numpy as np

array = np.random.randint(1,101,size=15)
print("Este es el array con 15 números aleatorios: ", array)
#multiplicar valores con bucle
total_multi = 1
for numeros in array:
    total_multi = total_multi *  numeros
print("La multiplicación de todos los números con bucle es : ", total_multi)
#multiplicar valores con numpy

total_mul_num = np.prod(array)
print("La multiplicación de todos los números con función numpy es : ", total_mul_num)
```

6. Haz lo mismo con la multiplicación elemento por elemento. Usa un operador y
después con una función de numpy
(Pista: busca en google que función de numpy hace esto) 

```python
import numpy as np

array = np.random.randint(1,101,size=15)
print("Este es el array con 15 números enteros aleatorios: ", array)
print("---")

array_dec = np.random.uniform(1,101,size=15)
array_dec = np.round(array_dec,2)
print("Este es el array con 15 números decimales aleatorios: ", array_dec)
print("---")

mul_array = 1
for num in array:
    mul_array = mul_array * num
print("Multiplicación array con bucle: ", mul_array)

mul_array_dec = 1
for num in array_dec:
    mul_array_dec = mul_array_dec * num
print("Multiplicación array decimal con bucle: ", mul_array_dec)

mul_array_np = np.prod(array)
print("Multiplicación de valores con np: ",mul_array_np)

mul_arrayDec_np = np.prod(array_dec)
print("Multiplicación de valores decimales con np: ",mul_arrayDec_np)
```

7. Encuentra el valor más alto del primer array que has creado.

8. Calcula la media (mean), la mediana (median) y al deviación estandar (standard
deviation) de los arrays (Nota: No nos importa el significado matemático de estos
valores, lo importante es que encuentres que función de numpy necesitas. Puedes
hacer la búsqueda en castellano o en inglés, aunque en inglés muchas veces suele
haber más resultados.)

```python
import numpy as np

array = np.random.randint(1,101,size=15)
print("Este es el array con 15 números aleatorios: ", array)
print("---")

valor_medio = np.mean(array)
print("Este es el valor medio de todo el array", valor_medio)
print("---")

valor_central_enOrden = np.median(array)
print("Este es el valor con función median",valor_central_enOrden)
print("---")

sd = np.std(array)
print("Este es el valor de la desviación estandar: ", sd)
```
